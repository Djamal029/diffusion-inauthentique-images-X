## Perpetual Hashing des images

Pourquoi cela est utile ? En effet il peut avoir des images qui sont légèrement modifiées, renommées, recadrées. On pourra donc calculer un Hash pour toutes images et les regrouper par Hash au lieu de image_id. Ce pourrait être plus précis.
Les noms d’images ne garantissent pas du tout qu’une image est identique.
Dans des scénarios réels de mésinformation, ils modifieraient presque toujours l’image (entk moi à leur place après connaissance du sijet).

* Les images ont été renommées : 
    - Twitter/X renomme souvent les images selon son propre hash interne.
    - Deux images identiques peuvent donc avoir des noms différents :

    ```
    A.jpg   ≠   BGWQk78XAAA.jpg
    ```
    Et parfois deux fichiers différents ont le même nom (rare, mais possible si l’export est buggué).
* Les images ont été compressées :  Si un utilisateur télécharge l’image puis la republie, elle sera recompressée => pixels changent.

* Les images ont été recadrées : Je rogne l'images et je republies.

* Les images ont un filtre, un sticker, un texte ajouté : Le nom ne change pas mais le contenu oui - ou inversément.

Même si les fichiers ont un nom identique dans ton extraction actuelle,on peut pas garantir qu'ils représentent bien :

- la même image visuelle

- ou une version modifiée

- ou une autre image renommée identiquement


Step 1 : utiliser le nom pour regrouper les reposts exacts (On a déjà fait mais pas ronuste)
- Très simple
- Très rapide
- Graphes coordination évidents

Step : ajouter pHash pour les clusters plus subtils
 - Détection de campagnes plus intelligentes

````python
import imagehash
from PIL import Image, ImageOps


def pipeline_phashing(image_path):
    img = Image.open(image_path)

    # corriger l’orientation EXIF
    img = ImageOps.exif_transpose(img)

    # conversion en niveaux de gris
    img = img.convert("L")

    # autocontrast léger (OPTIONNEL mais utile)
    img = ImageOps.autocontrast(img)

    # perceptual hash
    return imagehash.phash(img)


hash_bird_1 = imagehash.colorhash(image=Image.open("oiseau.jpg"))
hash_bird_2 = imagehash.colorhash(image=Image.open("oiseau_2.jpg"))
hash_al = imagehash.colorhash(image=Image.open("American_Alligator.jpg"))
````
